### What is Machine Learning ?
#### It is the process by which computers are given the ability to learn to make decisions from data w/o being explicitlity programmed.
#### Ex- E-mail is spam or not

### What is Unsupervised Learning ?
#### Uncovering hidden patterns from unlabelled data
#### Ex- Grouping customers into distinct catergories (Clustering)

### What is Supervised Learning ?
#### The predicted values are known
#### Aim: Predict the target values of unseen data, given from features
#### uses features to predict the value of target variable
#### Ex- predicting position of a player based on theri points per game

### Typer of supervised learning
#### **Classification** : Target Value consists of categories like binary classification
#### **Regression** : Target variable is continuous

### Requirements before performing Supervised Learning

*   No missing value
*   Data in Numeric format
*   Data stored in pandas DataFrame or Numpy array

```python
# Import the module
from sklearn.model_selection import train_test_split

X = churn_df.drop("churn", axis=1).values
y = churn_df["churn"].values

# Split into training and test sets
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=42, stratify=y)
knn = KNeighborsClassifier(n_neighbors=5)

# Fit the classifier to the training data
knn.fit(X_train, y_train)

# Print the accuracy
print(knn.score(X_test, y_test))

# Create neighbors
neighbors = np.arange(1, 13)
train_accuracies = {}
test_accuracies = {}

for neighbor in neighbors:

	# Set up a KNN Classifier
	knn = KNeighborsClassifier(n_neighbors=neighbor)

	# Fit the model
	knn.fit(X_train, y_train)

	# Compute accuracy
	train_accuracies[neighbor] = knn.score(X_train,y_train)
	test_accuracies[neighbor] = knn.score(X_test, y_test)
print(neighbors, '\n', train_accuracies, '\n', test_accuracies)


# Add a title
plt.title("KNN: Varying Number of Neighbors")

# Plot training accuracies
plt.plot(neighbors, train_accuracies.values(), label="Training Accuracy")

# Plot test accuracies
plt.plot(neighbors, test_accuracies.values(), label="Testing Accuracy")

plt.legend()
plt.xlabel("Number of Neighbors")
plt.ylabel("Accuracy")

# Display the plot
plt.show()
```
