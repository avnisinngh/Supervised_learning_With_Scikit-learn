### What is Machine Learning ?
#### It is the process by which computers are given the ability to learn to make decisions from data w/o being explicitlity programmed.
#### Ex- E-mail is spam or not

### What is Unsupervised Learning ?
#### Uncovering hidden patterns from unlabelled data
#### Ex- Grouping customers into distinct catergories (Clustering)

### What is Supervised Learning ?
#### The predicted values are known
#### Aim: Predict the target values of unseen data, given from features
#### uses features to predict the value of target variable
#### Ex- predicting position of a player based on theri points per game

### Typer of supervised learning
#### **Classification** : Target Value consists of categories like binary classification
#### **Regression** : Target variable is continuous

### Requirements before performing Supervised Learning

*   No missing value
*   Data in Numeric format
*   Data stored in pandas DataFrame or Numpy array
